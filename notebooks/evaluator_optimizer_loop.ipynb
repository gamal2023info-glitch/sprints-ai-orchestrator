# Evaluator-Optimizer loop (Assessor Agent)
Runs the grader + LLM-judge loop on sample submissions and records iterations, proving the 3-refinement cap.

````python
from dotenv import load_dotenv; load_dotenv()
from src.agents.orchestrator import build_graph, ask
graph = build_graph()
````

````python
good = '''```python\ndef is_even(n):\n    """Return True if n is even."""\n    return n % 2 == 0\nprint(is_even(4), is_even(7))\n```'''
out = ask(graph, 'SPR-1001', 'Grade:\n'+good, assignment='is_even', expected_output='True False')
print(out['grading'], out['judge'], out['refinement_count'])
````

````python
buggy = '''```python\ndef is_even(n):\n    return n % 2\nprint(is_even(4), is_even(7))\n```'''
out = ask(graph, 'SPR-1002', 'Grade:\n'+buggy, assignment='is_even', expected_output='True False')
print(out['grading'], out['judge'], out['refinement_count'])
assert out['refinement_count'] <= 3
````

````python
# Security: learner code that tries to touch the OS must be blocked, never executed
out = ask(graph, 'SPR-1003', 'Grade:\n```python\nimport os\nprint(os.listdir("/"))\n```')
print(out['execution'])
````
